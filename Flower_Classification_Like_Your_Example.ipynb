```python
import tensorflow as tf
import json
```

```python
# Load images
data = tf.keras.utils.image_dataset_from_directory(
    "train",
    image_size=(64, 64),
    batch_size=2
)

print(data.class_names)
```

```python
# CNN model (trained from scratch - no transfer learning)
model = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(64, 64, 3)),
    tf.keras.layers.Rescaling(1./255),
    tf.keras.layers.Conv2D(16, (3,3), activation="relu"),
    tf.keras.layers.MaxPooling2D(),
    tf.keras.layers.Flatten(),
    tf.keras.layers.Dense(5, activation="softmax")
])

# Compile
model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)
```

```python
# Train
model.fit(data, epochs=50)
model.summary()
```

```python
# Prediction
# Replace this path with an image from your dataset or a new flower image.
img = tf.keras.utils.load_img(
    r"Test\3530500952_9f94fb8b9c_m.jpg",
    target_size=(64,64)
)

x = tf.keras.utils.img_to_array(img)
x = tf.expand_dims(x, 0)

p = model.predict(x)
print(data.class_names[p.argmax()])
```

```python
# Save model
model.save("FLOWER.keras")

# Save class names to a JSON file
with open("classes.json", "w") as f:
    json.dump(data.class_names, f)
```

```python
# Load saved model and class names
with open("classes.json", "r") as f:
    class_names = json.load(f)

m = tf.keras.models.load_model("FLOWER.keras")

# Replace this path with the image you want to classify
img = tf.keras.utils.load_img(
    r"Test\3171577977_8608282f04_m.jpg",
    target_size=(64,64)
)

x = tf.keras.utils.img_to_array(img)
x = tf.expand_dims(x, 0)

p = m.predict(x)
print(p)
print(class_names[p.argmax()])
```
